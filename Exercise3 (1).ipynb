## 5. Groupby & Aggregation `(13 pts)`

Compute the average Attack per primary type.

```python
# put your answer here
df.groupby('Type 1')['Attack'].mean().sort_values(ascending=False)
```

Compute the maximum HP per generation.


```python
# put your answer here
df.groupby('Generation')['HP'].max()
```

Compute the total number of Pokémon per primary type.

```python
# put your answer here
df.groupby('Type 1').size().sort_values(ascending=False)
# Alternative (same as 4.1):
# df['Type 1'].value_counts()
```

For each type, compute:

- mean Attack

- mean Defense

- mean Speed

```python
# put your answer here=
df.groupby('Type 1')[['Attack', 'Defense', 'Speed']].mean().sort_values(by='Attack', ascending=False)
```

For each generation, compute:

- count

- mean Total

- number of Legendary Pokémon (hint: use sum on Boolean)

```python
# put your answer here
df.groupby('Generation').agg(
    Count=('Name', 'size'),
    Mean_Total=('Total', 'mean'),
    Legendary_Count=('Legendary', 'sum')
).sort_values(by='Mean_Total', ascending=False)
```

Which type combination (Type 1 + Type 2) has the highest average Attack?

```python
# put your answer here
df['Type 2'] = df['Type 2'].fillna('None')
mean_attack_combo = df.groupby(['Type 1', 'Type 2'])['Attack'].mean()


highest_attack_combo = mean_attack_combo.idxmax()
max_mean_attack = mean_attack_combo.max()

print(f"Highest Average Attack Combo: {highest_attack_combo} with mean Attack: {max_mean_attack:.2f}")
```

Which generation has the highest proportion of Legendary Pokémon?

```python
# put your answer here
gen_stats = df.groupby('Generation').agg(
    Legendary_Count=('Legendary', 'sum'),
    Total_Count=('Legendary', 'size')
)

# 2. Calculate the proportion
gen_stats['Legendary_Proportion'] = gen_stats['Legendary_Count'] / gen_stats['Total_Count']

# 3. Find the generation with the highest proportion
gen_stats['Legendary_Proportion'].sort_values(ascending=False).head(1)
```
